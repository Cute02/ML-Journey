```python
def most_frequent_emotions(detections):
    dict_emotions={}
    for i in range(len(detections)):
        if detections[i][1]>=0.3:
            if detections[i][0] in dict_emotions:
                dict_emotions[detections[i][0]]+=1
            else:
                dict_emotions[detections[i][0]]=1
    frequent=max(dict_emotions.values())
    for k, v in dict_emotions.items():
        if frequent==v:
            return k

print(most_frequent_emotions([("joy", 0.8), ("sadness", 0.2), ("joy", 0.6), ("anger", 0.5)]))
```

```python
def celsius_to_fahrenheit(C):
    F=(C * 9/5) + 32
    return F
print(celsius_to_fahrenheit(37))
```

```python
def classify_number(n):
    if n<0:
        return "Negative"
    elif n==0:
        return "Zero"
    elif n>0:
        if n%2==0:
            return "Positive Even"
        else:
            return "Positive Odd"
print(classify_number(-5))
print(classify_number(0))
print(classify_number(8))
print(classify_number(7))   
```

```python
# Using built in functions
def unique_sorted_data(data):
    return list(sorted(set(data)))
print(unique_sorted_data([5, 2, 8, 2, 5, 1]))
```

```python
# From scratch
def unique_from_scratch(data):
    unique=[]
    for i in data:
        if i not in unique:
            unique.append(i)
    for j in range(len(unique)):
        for k in range(j+1, len(unique)):
            if unique[j]>unique[k]:
                unique[k], unique[j]=unique[j], unique[k]
    return unique
print(unique_from_scratch([1, 2, 3, 7, 9]))
```

```python
def apply_discount(prices, discount_percent):
    discounted_list=[price - (discount_percent/100) * price for price in prices]
    return discounted_list
print(apply_discount([100, 200, 50], 10))
```

```python
import json

def save_and_load_scores(scores_dict, f_name):
    with open(f_name, "w") as fw:
        json.dump(scores_dict, fw)
    with open(f_name, "r") as fr:
        data=json.load(fr)
        return data


print(save_and_load_scores({"joy": 0.8, "sadness": 0.2}, "scores.json"))
```

```python
def analyze_conversation(messages):
    class ListEmpty(Exception):
            pass
    try:
        if len(messages)==0:
             raise ListEmpty
    except ListEmpty:
         return {}
    else:
        dict_freq={}
        for i in range(len(messages)):
            if messages[i]["confidence"]>=0.3:
                if messages[i]["emotion"] in dict_freq:
                    dict_freq[messages[i]["emotion"]]+=1
                else:
                    dict_freq[messages[i]["emotion"]]=1
        return dict_freq
print(analyze_conversation([
    {"text": "I feel great", "emotion": "joy", "confidence": 0.9},
    {"text": "so tired", "emotion": "sadness", "confidence": 0.2},
    {"text": "amazing day", "emotion": "joy", "confidence": 0.7},
    {"text": "ugh", "emotion": "anger", "confidence": 0.4},
]))
```
