```python
import pandas as pd
```

```python
data = {
    "emotion": ["joy", "sadness", "joy", "anger", "sadness", "joy"],
    "confidence": [0.9, 0.4, 0.6, 0.8, 0.3, 0.75]
}
df=pd.DataFrame(data)
print(df[(df["emotion"]=="joy") & (df["confidence"]>0.65)])
```

```python
print(df.groupby("emotion")["confidence"].mean())
print(df["emotion"].value_counts())
```

```python
df2 = pd.DataFrame({
    "user": ["A", "B", "C", "D"],
    "score": [85, None, 90, None]
})
print(df2["score"].isnull().sum())
```

```python
df2["score"]=df2.fillna(df2["score"].mean())
```

```python
df = pd.DataFrame({"user": ["A", "B", "C", "D"], "age": [25, -5, 130, 40]})
valid_median=df.loc[(df["age"]>=0) & (df["age"]<=120), "age"].median()
df.loc[(df["age"]<0) | (df["age"]>120), "age"]=valid_median

print(df)
```
