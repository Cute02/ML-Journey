```python
A = [2, 4, 6]
B = [1, 3, 5]
def dot_product(a, b):
    total=0
    for i in range(len(a)):
        total+=a[i]*b[i]
    return total
print(dot_product(A, B))
```

```python
arr=[[1, 2, 3], [4, 5, 6]]

def transpose(a):
    result=[[0 for y in range(len(a))]for x in range(len(a[0]))]
    for i in range(len(a)):
        for j in range(len(a[0])):
            result[j][i]=a[i][j]
    return result
print("Transpose", transpose(arr))
```
