```python
def mean(data):
    length=len(data)
    total=0
    for i in data:
        total+=i
    return  total/length
print("Mean of the data: ", mean([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
def mode(data):
    dict_freq={}
    for i in data:
        if i in dict_freq:
            dict_freq[i]+=1
        else:
            dict_freq[i]=1
    for k, v in dict_freq.items():
        if max(dict_freq.values())==v:
            return k
print(mode([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
def median(data):
    length=len(data)
    mid=length//2
    for i in range(len(data)):
        for j in range(len(data)):
            if data[i]<data[j]:
                data[i], data[j]=data[j], data[i]
    if length%2==0:
        return (data[mid]+data[mid-1])/2
    else:
        return data[mid]
   
print(median([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
def variance(data):
    total=0
    avg=mean(data)
    for i in data:
        total+=(i-mean(data))**2
    my_variance=total/len(data)
    return my_variance
print(variance([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
def std_dev(data):
    return variance(data)**0.5
print(std_dev([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
data=[3, 7, 7, 2, 9, 7, 4]
def my_mean(data):
    return sum(data)/len(data)
print(my_mean(data))
```

```python
def my_median(data):
    length = len(data)
    mid=length//2
    for i in range(len(data)):
        for j in range(i+1,len(data)):
            if data[i]>data[j]:
                data[i], data[j]=data[j], data[i]
    if length%2==0:
        return (data[mid]+data[mid-1])/2
    else:
        return data[mid]
print(my_median(data))
```

```python
def my_mode(data):
    dict_freq={}
    for i in data:
        if i in dict_freq:
            dict_freq[i]+=1
        else:
            dict_freq[i]=1
    max_value=max(dict_freq.values())
    for key,value in dict_freq.items():
        if value==max_value:
            return key
print(my_mode(data))
```
