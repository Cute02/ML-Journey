```python
data = {
    "emotion": ["joy", "sadness", "joy", "anger", "sadness", "joy"],
    "confidence": [0.9, 0.4, 0.6, 0.8, 0.3, 0.75]
}

df=pd.DataFrame(data)
counts=df["emotion"].value_counts()
plt.bar(counts.index, counts.values)
plt.xlabel("Emotion")
plt.ylabel("Count")
plt.show()
```

```python
days = [1, 2, 3, 4, 5, 6, 7]
joy_scores = [0.6, 0.7, 0.5, 0.8, 0.9, 0.6, 0.75]
sadness_scores = [0.3, 0.2, 0.4, 0.1, 0.15, 0.35, 0.2]

plt.subplot(1,2,1)
plt.plot(days, joy_scores )
plt.title("Joy Over time")

plt.subplot(1, 2, 2)
plt.plot(days, sadness_scores)
plt.title("Sadness Over time")

plt.suptitle("Emotion overtime")
plt.tight_layout()
plt.show()
```

```python
df4 = pd.DataFrame({
    "emotion": ["joy", "sadness", "joy", "anger", "sadness", "joy", "anger"],
    "confidence": [0.9, 0.4, 0.6, 0.8, 0.3, 0.75, 0.65]
})
grouped = df4.groupby("emotion")["confidence"].mean()
plt.bar(grouped.index, grouped.values)
plt.xlabel("Emotion")
plt.ylabel("Average Confidence")
```

```python
"""
grouped is a Series. A grouped Series has two parts:

.index — the labels (here, the emotion names: joy, sadness, anger)
.values — the actual numbers (the average confidence for each emotion)
"""
```

```python
grouped_emotion = df4.groupby("emotion")["confidence"].mean()
plt.bar(grouped_emotion.index, grouped_emotion.values)
plt.xlabel("Emotion")
plt.ylabel("Average Confidence")
plt.title("Overall emotion chart")
plt.show()
```
