```python
arr = np.array([100, 200, 300, 400, 500])
view_arr = arr[-3:].view()

view_arr[0] = 999
print("Shared array: ", view_arr)
print("Original array: ", arr)
```

```python
arr = np.array([100, 200, 300, 400, 500])
copied = arr.copy()
copied[0] = 999
print("Copied: ", copied)
print("Original array: ", arr)
```

```python
def sum_even_only(arr):
    total=0
    for x in np.nditer(arr):
        if x%2==0:
            total+=x
    return total
a=np.array([[1, 2, 3], [4, 5, 6]])
print(sum_even_only(a))
```

```python
arr=np.array([5, 12, 8, 20, 3, 17, 9, 25])
new=np.array_split(arr, 4)
for i in new:
    print(i, end=" ")
```

```python
mask=np.where(arr>10)
print(mask)
```
