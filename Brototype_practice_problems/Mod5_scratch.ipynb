```python
def matrix_scalar_multiply(matrix, scalar):
    result=[]
    for row in matrix:
        new_row=[]
        for value in row:
            new_row.append(value*scalar)
        result.append(new_row)
    return result
print(matrix_scalar_multiply(matrix_A, 4))
```

```python
def matrix_addition(matrix, scalar):
    result=[]
    for row in range(len(matrix)):
        new_row=[]
        for value in range(len(matrix[0])):
            new_row.append(value*scalar)
        result.append(new_row)
    return result

print(matrix_addition(matrix_A, matrix_B))
```

```python
def matrix_multiply(A, B): #A=(1,3) B=(3,1)
    rows_A=len(A) # 1
    cols_A=len(A[0]) # 3
    cols_B=len(B[0]) # 1

    result=[[0 for _ in range(cols_B) ] for _ in range(rows_A)]

    for i in range(rows_A):
        for j in range(cols_B):
            total=0
            for k in range(cols_A):
                total+=A[i][k]*B[k][j]
            result[i][j]=total
    return result

matrix_A = [[2, 3]]
matrix_B= [[4], [5]]


print(matrix_multiply(matrix_A, matrix_B))
```

```python
def magnitude(vector):
    sum_of_squares=0
    for i in vector:
        sum_of_squares+=i**2
    return sum_of_squares**0.5
vector_a=[6, 8, 12]
print(magnitude(vector_a))
```

```python
def unit_vector(vector):
    result=[]
    mag=magnitude(vector)
    for i in vector:
        result.append(i/mag)
    return result
print(unit_vector(vector_a))
```

```python
def cosine_similrity(v1, v2):
    mag1 = magnitude(v1) # sum of squares  of v1
    mag2 = magnitude(v2) # sum of squares  of v2
    dp=dot_product(v1, v2)
    cosine_value=dp/(mag1*mag2) 
    return cosine_value
print(cosine_similrity( [2, 0, 5], [1, 3, 2] ))
```

```python
ages = [20, 45, 60, 25, 80]
def normalize(vector):
    min_val=min(vector) #20
    max_val=max(vector) #80

    result=[]
    for x in vector:
        normalized_value=(x-min_val)/(max_val-min_val) # max-min= 60
        result.append(normalized_value)
    return result
print(normalize([10, 20, 30]))
```
