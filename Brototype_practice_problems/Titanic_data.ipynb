```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset('titanic')

print("=== 1. SHAPE ===")
print(df.shape, "\n")

print("=== 2. TYPES ===")
print(df.dtypes, "\n")

print("=== 3. TARGET ===")
print(df['survived'].value_counts(), "\n")

print("=== 4. MISSING VALUES (per column) ===")
print(df.isnull().sum()[df.isnull().sum() > 0], "\n")

print("=== 5. DISTRIBUTION (age, fare) ===")
print(df[['age', 'fare']].describe(), "\n")

print("=== 6. CORRELATION with target (numeric features only) ===")
print(df.corr(numeric_only=True)['survived'].abs().sort_values(ascending=False))

print("===7.Information===")
print(df.info())

print("===8.Statistical Information===")
print(df.describe())
```
