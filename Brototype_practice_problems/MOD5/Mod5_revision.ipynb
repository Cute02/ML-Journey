```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([35, 40, 50, 55, 65, 68])
plt.plot(hours, marks)
plt.xlabel("Hours studied")
plt.ylabel("Marks scored")
plt.title("Checking linearity: does this look like a straight line?")
plt.show()
```

```python
actual = np.array([35, 40, 50, 55, 65, 68])
predicted = np.array([33, 42, 48, 58, 63, 70])

residuals=actual - predicted
plt.plot(residuals, marker='o')
plt.axhline(0, color='red', linestyle='--')
plt.title("Residuals over time")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

predicted = np.array([10, 20, 30, 50, 80, 120])
residuals = np.array([1, -2, 2, -8, 15, -25])   # errors growing as predicted value grows

plt.plot(predicted, residuals)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Predicted value")
plt.ylabel("Residual (error)")
plt.title("Checking homoscedasticity: does error spread stay constant?")
plt.show()
```

```python
data = pd.DataFrame({
    "experience_years": [1, 3, 5, 8, 10, 15],
    "age": [23, 26, 29, 33, 36, 42],
    "salary": [30000, 40000, 55000, 70000, 85000, 110000]
})

correlation_matrix=data[["experience_years", "age"]].corr()
print(correlation_matrix)
```

```python
residuals = np.array([0.5, -0.3, 0.1, -0.8, 0.4, -0.2, 1.2, -0.6, 0.3, -0.1])

plt.hist(residuals, bins=6, edgecolor='black')
plt.title("Checking normality: do residuals form a bell shape?")
plt.show()

# Or a Q-Q plot (compares residuals to a theoretical normal distribution)
import scipy.stats as stats
stats.probplot(residuals, dist="norm", plot=plt)
plt.show()
```
