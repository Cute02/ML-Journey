```python
def matrix_add(A, B):
    addition_matrix=[]
    for i in range(len(A)):
        for j in range(len(A[0])):
            addition_matrix.append(A[i][j]+B[i][j])
    return addition_matrix
print(matrix_add([[2, 4], [6, 8]], [[10, 12], [14, 16]]))
```

```python
def scalar_multiply(A, k):
    scalar_multiplied_matrix=[]
    for i in range(len(A)):
        new_row=[]
        for j in range(len(A[0])):
            new_row.append(A[i][j]*k)
        scalar_multiplied_matrix.append(new_row)
    return scalar_multiplied_matrix

print(scalar_multiply([[2, 4], [6, 8]], 2))
```

```python
def matrix_multiply(A, B):
    rows_A=len(A)
    cols_B=len(B[0])
    cols_A=len(A[0])

    result=[[0 for x in range(cols_B)]for y in range(rows_A)]

    for i in range(rows_A):
        for j in range(cols_B):
            total=0
            for k in range(cols_A):
                total+=A[i][k]*B[k][j]
            result[i][j]=total           
    return result
A=[[2, 3], [4, 5]]
B=[[6, 7], [8, 9]]
print(matrix_multiply(A,B))
```

```python
def transpose(A):
    rows_A=len(A)
    cols_A=len(A[0])
    result=[[0]*rows_A for y in range(cols_A)]
    for i in range(rows_A):
        for j in range(cols_A):
            result[j][i]=A[i][j]
    return result
print(transpose([[2, 3], [4, 5], [6, 7]]))
```

```python

def dot_product(a, b):
    total=0
    for i in range(len(a)):
        total+=a[i]*b[i]
    return total

def magnitude(v):
    total=0
    for i in range(len(v)):
        total+=v[i]**2
    return total**0.5

a=[2, 3]
b= [4, 5]
mag_a=magnitude(a)
mag_b=magnitude(b)

cosine_similarity=dot_product(a,b)/(mag_a*mag_b)
print(cosine_similarity)
```

```python
def matrix_multiply(A, B):
    rows_A, cols_A=len(A), len(A[0])
    rows_B, cols_B= len(B), len(B[0])
    
    if cols_A!=rows_B:
        raise ValueError("Columns of A should be eqaual to Rows of B")
    
    result=[[0 for y in range(cols_B)]for x in range(rows_A)]
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j]+=A[i][k]*[k][j]
    return result
print(matrix_multiply([[2, 3], [4, 5]], [[2], [3]]))
```

```python
def matrix_scalar(A, k):
    result=[[0 for y in range(len(A[0])) ]for x in range(len(A))]
    for i in range(len(A)):
        for j in range(len(A[0])):
            result[i][j]=A[i][j]*k
    return result
print(matrix_scalar([[2, 3], [4, 5]], 2))
```

```python
def transform_row(row, weights, bias):
    total=0
    for i in range(len(row)):
        total+=row[i]*weights[i]
    return total + bias
def transform_data(data, weights, bias):
    return [transform_row(row, weights, bias ) for row in data]

print(transform_data([[2, 3], [4, 5]],[6, 7] ,10))
```

```python
def slope(guess, target):
    return 2*(guess- target)
def gradient_descent(start, target, lr, steps):
    x=start
    history =[x]
    for i in range(steps):
        x=x-lr*slope(x, target)
        history.append(x)
    return history

print(gradient_descent(6, 7, 0.3, 10))
```
