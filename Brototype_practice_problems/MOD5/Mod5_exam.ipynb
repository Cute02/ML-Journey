```python
vector1 = [[2, 4, 6], [7, 11, 13], [8, 9, 10]]
vector2 = [[1, 3, 5], [7, 11, 13], [2, 4, 6]]
def dot_product(v1, v2):
    total=0
    for i in range(len(v1)):
        total+=v1[i]*v2[i]
    return total
rows_v1=len(vector1)
cols_v2=len(vector2[0])
result=[[0 for y in range(cols_v2)] for x in range( rows_v1)]
for i in range(rows_v1):
    for j in range(cols_v2):
        result[i][j]+=dot_product(vector1[i][j], vector2[j][i])
print("Dot product: ", dot_product(vector1, vector2))
```

```python
def transpose(v):
    result=[[0 for y in range(len(v))]for x in range(len(v[0]))]
    for i in range(len(v)):
        for j in range(len(v[0])):
            result[j][i]=v[i][j]
    return result
print(transpose([[1, 2], [3, 4], [5, 6]]))
```

```python
a=[
    [1, 2],
    [3, 4]
]
b=[
    [5, 6],
    [7, 8]
]
def matrix_multiply(A, B):
    rows_A, cols_A=len(A), len(A[0])
    cols_B=len(B[0])
    result=[[0 for y in range(cols_B)] for x in range(rows_A)]
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j]+=A[i][k]*B[k][j]
    return result
print("Matrix Multiplication: ", matrix_multiply(a, b))
```
