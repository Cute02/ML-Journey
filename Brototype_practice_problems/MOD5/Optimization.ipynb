```python
import numpy as np

hours=np.array([1, 2, 3, 4, 5]) # x
marks=np.array([35, 40, 50, 55, 65]) # y



m,c=0,0
learning_rate=0.01
n=len(hours)

for step in range(1000):
    pred=m*hours+c
    error=pred-marks

    dm=(2/n)*np.sum(error*hours)
    dc=(2/n)*np.sum(error)

    m-=learning_rate*dm
    c-=learning_rate*dc
    

print(f"Best fit line marks= {m:.2f}* hours + {c:.2f} ")
print(f"Prediction for  6 hours: {m*6+c:.2f}.")
```

```python
neha = np.array([5, 1, 3])
friend = np.array([4, 2, 3])

dot_Product=np.dot(neha, friend)
mag_neha=np.linalg.norm(neha)
mag_friend=np.linalg.norm(friend)
cosine_simi=dot_Product/(mag_neha*mag_friend)
print(f"{cosine_simi:.3f}")
```

```python
house_A = np.array([12, 5])
house_B = np.array([15, 9])
eucli_dist=np.linalg.norm(house_A-house_B)
print(f"{eucli_dist:.3f}")
```

```python
ages = np.array([5, 25, 45, 65, 80])
scaled_ages=(ages-ages.min())/(ages.max()-ages.min())
print(f"Scaled ages= {scaled_ages}")
```
