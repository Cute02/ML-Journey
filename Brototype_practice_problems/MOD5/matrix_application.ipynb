```python
people_raw = [
    {"oiliness": "high", "redness": "medium", "dryness": "low"},
    {"oiliness": "low", "redness": "low", "dryness": "high"},
    {"oiliness": "medium", "redness": "high", "dryness": "medium"}
]
level_map = {"low": 0.1, "medium": 0.5, "high": 0.9}
```

```python
def represent_as_vector(row, mapp):
    return [ mapp[row[key]] for key in row]
```

```python
def build_feature_matrix(data, mapp):
    return[represent_as_vector(row, mapp) for row in data]
new_data=build_feature_matrix(people_raw, level_map)
```

```python
def normalize(data):
    smallest=min(data)
    biggest=max(data)
    result=[]
    for i in data:
        normalized_value=(i-smallest)/(biggest-smallest)
        result.append(normalized_value)
    return result
```

```python
def normalize_matrix_column(data, col_index):
    column=[row[col_index] for row in data]
    normalized_colmn=normalize(column)
    for i in range(len(data)):
        data[i][col_index]= normalized_colmn[i]
    return data
new_data=normalize_matrix_column(new_data, 0)
```

```python
def matrix_multiply(A, B):
    result=[[0 *len(B[0])]for x in range(len(A))]
    for i in range(len(A)):
        for j in range(len(B[0])):
            for k in range(len(A[0])):
                result[i][j]+=A[i][k]*B[k][j]
    return result
```

```python
acne_weights = [[0.7], [0.6], [0.1]]
for col in range(3):
    feature_matrix=normalize_matrix_column(new_data, col)
predictions=matrix_multiply(feature_matrix, acne_weights)

print(predictions)
```

```python
def derivative(f, x, step=0.0001):
    return (f(x+step)-f(x))/step


def f(x):
    return (x-5)**2
def g(x):
    return x**3

print(derivative(f,10))
```

```python
def gradient_descent(f, start_x, learning_rate=0.1, steps=50):
    x=start_x
    for i in range(steps):
        grad=derivative(f, x)
        x=x-learning_rate*grad
        return x 
best_x=gradient_descent(f, start_x=0)
print(best_x)
```
