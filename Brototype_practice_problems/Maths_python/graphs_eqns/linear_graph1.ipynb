```python
"""
Write the code to plot a simple line graph of x = [1,2,3], y = [2,4,6] 
using matplotlib, with axis labels and a title.
"""

import matplotlib.pyplot as plt
x = [1,2,3]
y = [2,4,6]
plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("USeless")
plt.show()
```

```python
import numpy as np

data=np.random.normal(loc=50, scale=10, size=1000)
```

```python
plt.hist(data, bins=30)
plt.show()
```

```python
indices=np.arange(1000)
plt.scatter(data, indices, alpha=0.5)
plt.show()
```

```python
plt.boxplot(data)
plt.show()
```
