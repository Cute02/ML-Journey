```python
import pandas as pd
```

```python
df=pd.read_csv('C:\\Brototype\\ML_learn\\ML_projects\\Data_Preparation\\DATASETS\\Titanic-Dataset.csv')
```

#### HANDLING MISSING OR DUPLICATE VALUES 

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df["Age"])
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
print(df["Embarked"])
```

```python
df=df.drop("Cabin", axis=1)
```

```python
df["Fare_scaled"]=(df["Fare"]-df["Fare"].min())/(df["Fare"].max()-df["Fare"].min())
print(df["Fare_scaled"].median())
```

#### LABEL ENCODING

```python
df["Sex_encoded"] = df["Sex"].map({"male": 0, "female": 1})
```

#### ONE HOT ENCODING

```python
df=pd.get_dummies(df, columns=["Embarked"], prefix="Embarked")
```

```python
print(df.info())
```

```python

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df["FamilySize"]
```

```python
df=df.merge(class_info, on="Pclass", how="left")
```
