```python
def f(x):
    return (x-5)**2
def numerical_derivative(f, x, h=0.0001):
    f_x=(f(x+h)-f(x))/h
    return f_x
print(numerical_derivative(f, 3))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def cost_function(x):
    return x**2
def analytical_derivative(x):
    return 2*x
x_space=np.linspace(-1, 5, 100)
y_space = cost_function(x_space)
current_x=4.0
current_cost=cost_function(current_x)
current_slope= analytical_derivative(current_x)

plt.figure(figsize=(8,5))
plt.plot(x_space, y_space, label='Cost function f(x)', color= 'blue')

plt.scatter(current_x, current_cost, color='red', zorder=5, label='Current State')

tangent_x = np.linspace(3, 5, 10)
tangent_y= current_slope*(tangent_x-current_x)+current_cost
plt.plot(tangent_x, tangent_y, 'r--', label=f'Tangent slope ({current_slope})')

plt.title("Visualizing Functions & Slopes in ML Optimization")
plt.xlabel("Parameter Weight (x)")
plt.ylabel("Error/Cost (f(x))")
plt.grid(True)
plt.legend()
plt.show()
```

```python
def gradient_descent(start_x, learning_rate, steps):
    x=start_x
    for i in range(steps):
        gradient=numerical_derivative(f, x, learning_rate)
        x=x-learning_rate*gradient
    return x
print(gradient_descent(start_x=0, learning_rate=0.1, steps=50))
```
