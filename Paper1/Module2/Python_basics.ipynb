```python
# Prime no

n=10
for i in range(2, n):
    isPrime=0
    for j in range(2, i+1):
        if i%j==0:
            isPrime=1
            break
    if isPrime==0:
        print(i)
```

```python
def dot_product(v1, v2):
    total_row=0
    for i in range(len(v1)):
        total_row+=v1[i]*v2[i]
    return total_row
v1 = [
    [1, 2],
    [3, 4]
]

v2 = [
    [5, 6],
    [7, 8]
]
total=0
for row in range(len(v1)):
    total+=dot_product(v1[row], v2[row])
print(total)
```
