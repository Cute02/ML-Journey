```python
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Simulate your historical credit card transactions (in Rupees)
# Most purchases are around ₹1,000, with a few slightly higher ones
history = np.array([[800], [1200], [950], [1100], [1050], [1500], [700], [900]])

# 2. Initialize the Scikit-Learn Scaler and "Learn" the Mean & Std Dev
scaler = StandardScaler()
scaler.fit(history)

# Extract the learned metrics to see what the machine saw
mean = scaler.mean_[0]
std_dev = np.sqrt(scaler.var_[0]) # Square root of variance = Std Dev

print(f"--- Your Spending Profile ---")
print(f"Calculated Mean: ₹{mean:.2f}")
print(f"Calculated Std Dev: ₹{std_dev:.2f}")
print(f"Fraud Trigger Boundary (Mean + 3*StdDev): ₹{mean + 3*std_dev:.2f}\n")

# 3. Incoming new swipes to test
# Swipe 1: Regular grocery trip (₹1,300)
# Swipe 2: Sudden massive luxury purchase (₹12,000)
new_swipes = np.array([[300], [12000]])

# 4. Transform raw values into Z-scores (Standard Deviations away from Mean)
z_scores = scaler.transform(new_swipes)

# 5. Check if the transaction passes the 3-Standard-Deviation rule
print(f"--- Real-Time Transaction Screening ---")
for i, swipe in enumerate(new_swipes):
    amount = swipe[0]
    z_score = z_scores[i][0]
    
    print(f"Swipe Amount: ₹{amount}")
    print(f"Distance from your average: {z_score:.2f} Standard Deviations")
    
    if abs(z_score) > 3.0:
        print("❌ STATUS: FLAGGED AS FRAUD! (Transaction Blocked)\n")
    else:
        print("✅ STATUS: APPROVED\n")
```
