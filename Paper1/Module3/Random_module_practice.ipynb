```python
import numpy as np
```

```python
np.random.seed(42)
random_values=np.random.normal(loc=100, scale=15, size=1000)
print(random_values[:5])
```

```python
def simulate_emotions(n):
    np.random.seed(1)
    return np.random.choice(["joy", "sadness", "anger"], n,  p=[0.6, 0.25, 0.15])
     
print(simulate_emotions(10))
```

```python
data = np.array([1, 2, 3, 4, 5])
print(np.random.permutation(data))
print(data)
```

```python
def simulate_responses(num_users, num_prompts, response_rate):
    return np.random.binomial(num_prompts, response_rate, num_users)
print(simulate_responses(5,20,0.7))
```
