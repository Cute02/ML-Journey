```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, f1_score

```

```python
cancer_data= load_breast_cancer()
X= pd.DataFrame(cancer_data.data, columns= cancer_data.feature_names)
y= cancer_data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
temp_scaler = StandardScaler()
X_train_scaled = temp_scaler.fit_transform(X_train)
temp_pca = PCA()

temp_pca.fit(X_train_scaled)
```

```python
cumulative_variance = np.cumsum(temp_pca.explained_variance_ratio_)
```

```python
n_components_90 = np.argmax(cumulative_variance>=90)+1
```

```python
pipeline = Pipeline([('scaler', StandardScaler()),
                      ('pca', PCA(n_components_90)), 
                      ('classifier', LogisticRegression(C=1.0))])
```

```python
pipeline.fit(X_train, y_train)
```
