```python
import numpy as np
```

```python
class LinearRegressionGD:
    def __init__(self, lr=0.01, epochs=1000):
        self.lr, self.epochs = lr, epochs

    def fit(self, X, y): # X= feature matrix , y= target vector
        n, m = X.shape
        self.w, self.b = np.zeros(m), 0.0 # sets weights and bias to 0
        for _ in range(self.epochs):
            error = (X@ self.w +self.b) - y # X*m +b -target to check how much error
            self.w -=self.lr*(2/n) * (X.T @error)
            self.b -= self.lr *(2/n)* error.sum()
            return self
    def predict(self, X):
        return X@self.w +self.b
```
