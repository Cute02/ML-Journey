```python
from sklearn.datasets import load_iris
iris=load_iris()
X=iris.data
y=iris.true_labels if hasattr(iris, 'true_labels') else iris.target
```

#### Supervised model


```python
from sklearn.tree import DecisionTreeClassifier

clf=DecisionTreeClassifier()
clf.fit(X, y)
print("Supervised prediction for first flower: ", clf.predict([X[0]]))
```

```python
from sklearn.cluster import KMeans

kmeans= KMeans(n_clusters=3, random_state=0, n_init=10)
kmeans.fit(X)
print("Unsupervised prediction for first flower: ",kmeans.labels_[0])
```
