```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame({
    "income":  [45000, 52000, np.nan, 61000, 45000],
    "city":    ["Kochi", "kochi", "Kochi", "Trivandrum", "Kochi"],
    "gender":  ["M", "Male", "M", "F", "M"],
})

print("Before:\n", df)

```

```python
print("Rows, columns: ", df.shape)
```

```python
df.isnull().sum()
```

```python
df['income'] = df['income'].fillna(df['income'].median())
```

```python
df['city'].unique()
```

```python
df['city'] = df['city'].str.strip().str.title()
```

```python
df['gender'] = df['gender'].replace({'Male': 'M'})
```

```python
df.duplicated()
```

```python
df= df.drop_duplicates()
```
